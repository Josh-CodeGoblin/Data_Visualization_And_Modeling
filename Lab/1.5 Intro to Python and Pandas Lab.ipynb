```python
import pandas as pd

df = pd.read_csv('../Data/archive/incidents.csv')

df.head()
```

```python
print(df.columns.tolist())
```

```python
df['incident_type'].value_counts()
```

```python
# Question 1: How many turbulence incidents are in the dataset?
# Filter the dataset to only show turbulence incidents
turbulence_df = df[df['incident_type'] == 'Turbulence']

# Count and display the total number of turbulence incidents
print("Number of turbulence incidents:", len(turbulence_df))
```

```python
Answer: There are 2,228 turbulence incidents in the dataset.
```

```python
# Question 2: Which region has the most turbulence incidents?
# Count turbulence incidents by region
airline_region_df = turbulence_df['region'].value_counts()

# Display the results
print(airline_region_df)
```

```python
Answer: North America has the most turbulence incidents with 642 incidents.
```

```python
# Question 3: During what phase of flight does turbulence happen the most?
# Count turbulence incidents by phase of flight
flight_phase_df = turbulence_df['phase_of_flight'].value_counts()

# Display the results
print(flight_phase_df)
```

```python
Answer: Turbulence happens the most during climb, with 765 incidents.
```

```python
# Question 4: What weather conditions are most common during turbulence incidents?
# Count turbulence incidents by weather condition
weather_df = turbulence_df['weather_conditions'].value_counts()

# Display the results
print(weather_df)
```

```python
Answer: VMC was the most common weather condition with 880 incidents.
```

```python
# Question 5: What is the average number of injuries during turbulence incidents?
# Calculate the average number of injuries
average_injuries = turbulence_df['injuries'].mean()

# Display the result
print("Average number of injuries:", average_injuries)
```

```python
Answer: The average number of injuries during turbulence incidents was 12.42.
```
